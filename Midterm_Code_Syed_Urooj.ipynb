Data Cleaning

```python
# Add a unique record number to distinguish each AED because some may be duplicates but still valid
df_aed['Record ID'] = range(1, len(df_aed) + 1)
```

```python
# Replace missing and "TBD" values in Division, Placement, and Postal Code with "Not Provided"
for column in ['Division', 'Placement', 'Postal Code']:
    df_aed[column] = df_aed[column].apply(
        lambda x: 'Not Provided'
        if pd.isna(x) or x == 'TBD'
        else x
    )

# Check remaining missing values
df_aed[['Division', 'Placement']].isna().sum()
```

```python
# Check remaining "TBD" values
(df_aed[['Division', 'Placement', 'Postal Code']] == 'TBD').sum()
```

```python
# Remove extra spaces from City values to make them consistent and combine duplicate values
df_aed['City'] = df_aed['City'].str.strip()

# Standardize City capitalization
df_aed['City'] = df_aed['City'].str.title()

# Check City values after removing extra spaces
df_aed['City'].unique()
```

```python
# Check all City values containing "Scar" to check for misspellings
df_aed[df_aed['City'].str.contains('Scar', case=False, na=False)]['City'].unique()
```

```python
# Check all City values containing "Eto" to check for misspellings
df_aed[df_aed['City'].str.contains('Eto', case=False, na=False)]['City'].unique()
```

```python
# Correct misspelled City values
df_aed['City'] = df_aed['City'].apply(
    lambda x: 'Scarborough' if x == 'Scarbourgh'
    else 'Etobicoke' if x == 'Etobicke'
    else x
)

# Check City values after correcting spelling
df_aed['City'].unique()
```

```python
# Save the cleaned dataset to a seperate CSV file
df_aed.to_csv('Midterm_Cleaned_Syed_Urooj.csv', index=False)
```

Basic Data Analysis

```python
# Load cleaned dataset for analysis
df_aed = pd.read_csv('Midterm_Cleaned_Syed_Urooj.csv')
```

```python
# Analysis 1: Count AED records by the division
aed_by_division = df_aed['Division'].value_counts()

print(aed_by_division)
```

```python
# Analysis 2: Identify cities with the lowest number of AED records
aed_by_city = df_aed['City'].value_counts()

# Find the lowest AED count
min_aed_count = aed_by_city.min()

# Display all cities with the lowest AED count
cities_min_aed = aed_by_city[aed_by_city == min_aed_count]

print(cities_min_aed)
```

```python
# Analysis 3: Identify the most common AED manufacturer
most_common_manufacturer = df_aed['AED Manufacturer'].mode()[0]

print('Most common AED manufacturer:', most_common_manufacturer)
```

```python
# Analysis 4: Identify unique AED manufacturers
aed_manufacturers = df_aed['AED Manufacturer'].unique()

print('AED manufacturers:', aed_manufacturers)
print('Number of manufacturers:', len(aed_manufacturers))
```

```python
# Analysis 5: Compare AED records from City of Toronto and Safe City Program
city_of_toronto_aeds = df_aed[df_aed['Organization'] == 'City of Toronto']
safe_city_aeds = df_aed[df_aed['Organization'] == 'Safe City Program']

print('City of Toronto AED records:', len(city_of_toronto_aeds))
print('Safe City Program AED records:', len(safe_city_aeds))
```

```python
# Analysis 6: Identify AED records with placement information not provided
missing_placements = df_aed[df_aed['Placement'] == 'Not Provided']

print('Number of AED records with placement not provided:', len(missing_placements))
print(missing_placements['Site'].to_string(index=False))
```

Higher-Level Data Analysis

```python
# Higher-Level Analysis 1: Calculate which organization has the highest proportion of records with placement information not provided

# Filter missing placements by organization
missing_placements_by_org = df_aed[df_aed['Placement'] == 'Not Provided'].groupby('Organization')

# Count of missing placements
missing_placements_by_org_count = missing_placements_by_org.size()

# Count total records
total_by_org = df_aed['Organization'].value_counts()

# Calculate percentage of missing Placement records by organization
missing_placement_percentage = (missing_placements_by_org_count / total_by_org) * 100

# Round missing Placement percentages to one decimal place
missing_placement_percentage = missing_placement_percentage.round(1)

# Find the organization with the highest percentage of missing Placement records
organization_with_highest_missing = missing_placement_percentage[
    missing_placement_percentage == missing_placement_percentage.max()]

print(organization_with_highest_missing)
```

```python
# Higher-Level Analysis 2: Identify sites with more AED records than the average site

# Count AED records for each site
aed_by_site = df_aed['Site'].value_counts()

# Calculate the average AED count
average_aed_count = aed_by_site.mean()

# Filter sites with AED counts above the average
sites_above_average = aed_by_site[aed_by_site > average_aed_count]

# Display the average and top 10 sites above the average
print('Average AED records per site:', round(average_aed_count, 2))
print('\nTop 10 sites with more AED records than the average:')
print(sites_above_average.head(10))


```
